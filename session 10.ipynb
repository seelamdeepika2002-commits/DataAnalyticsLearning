```python
      ######### Session – 10 ##########################
# Task 1: Memory Layout & Summation Speed Comparison

import numpy as np
import time
import sys

# Number of elements
n = 1_000_000

# Create a Python list
python_list = list(range(n))

# Create an equivalent NumPy array
numpy_array = np.arange(n, dtype=np.int64)

print("Number of elements:", n)
print("First 5 values:", python_list[:5])
print("Last 5 values:", python_list[-5:])

# Memory used by the list object itself
list_object_size = sys.getsizeof(python_list)

# Memory used by the individual integer objects inside the list
list_element_memory = sum(sys.getsizeof(x) for x in python_list)

# Estimated total memory used by the Python list
total_list_memory = list_object_size + list_element_memory

# NumPy array memory
# nbytes gives the size of the actual numerical data buffer
numpy_memory = numpy_array.nbytes

print(f"Python list object memory: {list_object_size:,} bytes")
print(f"Python list element memory: {list_element_memory:,} bytes")
print(f"Python list total estimated memory: {total_list_memory:,} bytes")
print(f"NumPy array memory: {numpy_memory:,} bytes")

memory_ratio = total_list_memory / numpy_memory

print(f"\nPython list memory: {total_list_memory:,} bytes")
print(f"NumPy array memory: {numpy_memory:,} bytes")
print(f"List uses approximately {memory_ratio:.2f} times more memory.")

# Measure Python list sum
start_time = time.perf_counter()

list_sum = sum(python_list)

list_time = time.perf_counter() - start_time


# Measure NumPy array sum
start_time = time.perf_counter()

numpy_sum = numpy_array.sum()

numpy_time = time.perf_counter() - start_time


print(f"Python list sum: {list_sum}")
print(f"NumPy array sum: {numpy_sum}")

print(f"\nTime using Python list: {list_time:.6f} seconds")
print(f"Time using NumPy array: {numpy_time:.6f} seconds")

print("----- Final Comparison -----")
print(f"Python list memory : {total_list_memory:,} bytes")
print(f"NumPy array memory : {numpy_memory:,} bytes")

print(f"\nPython list sum time : {list_time:.6f} seconds")
print(f"NumPy array sum time : {numpy_time:.6f} seconds")

if numpy_time > 0:
    print(f"\nNumPy sum is approximately {list_time / numpy_time:.2f} times faster.")
```

```python
#Task 3: Broadcasting Operations
#Analogy: Broadcasting in NumPy is like sending a single radio signal to multiple receivers. You
#will automatically “broadcast” a smaller array across a larger one according to NumPy’s rules.
#Instructions: Consider the following scenario: A company has daily base salaries for 3
#employees over 4 days in a 3×4 array base_pay. There is also a 1×4 row vector daily_bonus of
#bonuses (one bonus amount for each day). Use NumPy broadcasting to add the daily bonus to
#each employee’s base pay without writing an explicit loop.
#Example data to use:
#base_pay = np.array([[100, 100, 100, 100],
#[200, 200, 200, 200],
#[300, 300, 300, 300]])
#daily_bonus = np.array([10, 20, 30, 40])
#Compute total_pay = base_pay + daily_bonus using broadcasting.
#Explain briefly: Why this operation works (mention that 1×4 is broadcast to 3×4 per rules).
#Input: The arrays base_pay and daily_bonus as above.

import numpy as np

# Base salaries for 3 employees over 4 days
base_pay = np.array([
    [100, 100, 100, 100],
    [200, 200, 200, 200],
    [300, 300, 300, 300]
])

# Bonus for each day
daily_bonus = np.array([10, 20, 30, 40])

print("Base Pay:")
print(base_pay)

print("\nDaily Bonus:")
print(daily_bonus)

print("\nBase Pay shape:", base_pay.shape)
print("Daily Bonus shape:", daily_bonus.shape)
# NumPy automatically broadcasts daily_bonus across all 3 employees
total_pay = base_pay + daily_bonus

print("Total Pay:")
print(total_pay)
print("Result shape:", total_pay.shape)
print("Result data type:", total_pay.dtype)
```

```python
# Task 4: Vectorized Conditional Computation

import numpy as np

readings = np.array([-2, -1, 0, 1, 2])

results = np.where(readings >= 0, readings ** 2, readings ** 3)

print("Original readings:")
print(readings)

print("\nTransformed results:")
print(results)

# Explanation:
# np.where() performs the condition on the entire array without using an explicit Python loop.

# Negative values are cubed:
# -2^3 = -8
# -1^3 = -1
#
# Non-negative values are squared:
# 0^2 = 0
# 1^2 = 1
# 2^2 = 4
```
