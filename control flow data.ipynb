```python
#match-case
#match multiple inputs and given an output

day=input("enter the day of the week")

match day:
     case"mon":print("start of the week")
     case"tue":print("mid week")
     case"wed":print("mid week")
     case"thur":print("mid week")
     case"fri":print("end of the week")
     case"sat":print("end of the week")
     case"sun":print("end of the week")
     case  _:print("no such day exists")#default block
```

```python

# same as above conditions but minimal sentences

day=input("enter the day of the week")

match day:
     case"mon":print("start of the week")
     case"tue" | "wed" | "thur" :print("mid week")
     case"fri" |  "sat" | "sun":print("end of the week")
     case  _:print("no such day exists")#default block
```

```python
#matching with conditions

score=85

match score:
    case s if s>=90:
        print("grade A")
    case s if s>=60:
        print("grade B")
    case s if s>=40:
        print("grade C")
    case  _:
        print("fail")         
```

```python
#match with the tuple
#match case can unpack a tuple and check the shape of the data not just compare a single value
point=(7,6)
match point:
    case(0,0):print("on origin")
    case(0,y):print("on y axis at point",y)
    case(x,0):print("on x axis at point",x)
    case(x,y):print(f"point at {x},{y}")
```

```python
# Intermediate Practice Questions — Control Flow
# Conditionals, Match-Case, Truthy/Falsy
# 

# ------------------------------------------------------------
# Q1. Customer Discount Tier (Nested Conditionals + Truthy/Falsy)
#
# customer = {"name": "Riya", "orders": 12, "membership": "gold"}
#
# Write code that determines a discount percentage:
# - If "membership" key is missing or empty -> 0% discount
# - Else if membership is "gold" AND orders > 10 -> 20% discount
# - Else if membership is "gold" -> 10% discount
# - Else if orders > 5 -> 5% discount
# - Else -> 0% discount
#
# Use .get() to safely check for the membership key (don't
# assume it exists), and use truthy/falsy to check if it's empty.
#
# Expected output:
# 20% discount

customer = {"name": "Riya", "orders": 12, "membership": "gold"}
if not  customer .get("membership"):
    print("0% discount")
else:
    if customer.get("membership")=="gold"and customer.get("orders")>10:
        print("20% discount")
    elif customer.get("membership")=="gold":
        print("10% discount")
    elif customer.get("orders") >5:
        print("5%discount")
    else:
        print("0% discount")      

```
