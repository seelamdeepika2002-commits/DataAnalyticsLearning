```python
#Task 2: Immutable Records & Tuple Unpacking
#Objective: Use Tuples for fixed data structures and explore packing/unpacking mechanics.
#1. The Record: Create a tuple representing a professional employee record: (101, "Alice
#Smith", "Data Engineer", "London"
employee_record = (101, "Alice Smith", "Data Engineer", "London")

print(employee_record)
```

```python
#2. Unpacking: Unpack this tuple into four distinct
#  variables: emp_id, name, role, and location.
employee_record = (101, "Alice Smith", "Data Engineer", "London")

# Unpacking the tuple
emp_id, name, role, location = employee_record

print("Employee ID:", emp_id)
print("Name:", name)
print("Role:", role)
print("Location:", location)
```

```python
#3. Extended Unpacking: Create a larger tuple with 6 values. Use the "starred" expression
#(*) to unpack the first element into head, the last element into tail, and everything in
#between into a list called mid
# Create a tuple with 6 values
data = (101, "Alice", "Data Engineer", "London", "Python", 50000)

# Extended unpacking
head, *mid, tail = data

print("Head:", head)
print("Middle:", mid)
print("Tail:", tail)
```

```python
#Task 3: Deduplication and Set Hashability
#Objective: Utilize Sets to manage unique data and understand the constraints of hashable
#elements.
#1. Cleaning Data: You are given a raw list of user IDs: [1001, 1002, 1001, 1005,
#1003, 1002, 1008]. Convert this list into a set called unique_ids to automatically
#remove duplicates.
# Raw list of user IDs
user_ids = [1001, 1002, 1001, 1005, 1003, 1002, 1008]

# Convert list into a set to remove duplicates
unique_ids = set(user_ids)

print("Unique IDs:", unique_ids)
```

```python
#3. Set Growth: Use the .add() method to insert a new ID and the .update() method to
#add multiple IDs from another list: [1010, 1011].
unique_ids = {1001, 1002, 1005, 1003, 1008}

# Add a single new ID
unique_ids.add(1009)

# Add multiple IDs from another list
new_ids = [1010, 1011]
unique_ids.update(new_ids)

print("Updated Unique IDs:", unique_ids)
```
