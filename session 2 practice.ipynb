```python
#Task 2: Advanced I/O and Type Casting Pipeline
#Objective: Build a robust input-to-calculation pipeline using explicit type casting.
#1. User Acquisition: Use input() to collect:
#o hourly_rate (Expected: Float)
#o hours_worked (Expected: Int)
#o tax_bracket (Expected: Float, e.g., 0.2 for 20%)
#2. Casting: Cast these inputs directly into their respective types.
#3. Calculation: Calculate the gross_pay and the net_pay (Gross - (Gross * Tax)).
#4. Validation: Print the type() of each processed variable to ensure the casting was successful.

hourly_rate=float(input("enter the hourly rate"))
hours_worked=int(input("enter  how many hours worked"))
tax_bracket=float(input(" enter tax bracket"))
gross_pay=hourly_rate*hours_worked
net_pay=gross_pay-(gross_pay*tax_bracket)
print("gross pay:",gross_pay)
print("net pay:",net_pay)
print("Type of hourly_rate:", type(hourly_rate))
print("Type of hours_worked:", type(hours_worked))
print("Type of tax_bracket:", type(tax_bracket))





```

```python
#Task 3: Precision Engineering with F-Strings
#Objective: Generate a professional-grade text report using padding, alignment, and
#mathematical precision.
#1. Design Requirements: Using the data from Task 2, print a
#  "Salary Statement" with the
#following constraints:
# Header: A string "OFFICIAL PAYSLIP" centered in a 40-character
#  block using= as a filler.
# Labels: All labels (e.g., "Hourly Rate:") must be left-aligned with a width of 25.
# Values: All numerical values must be right-aligned, fixed to 3 decimal places,
#and preceded by a currency symbol.
#2. Visual Structure: Use string multiplication to create a divider line that exactly matches
#the width of your header.

hourly_rate = float(input("Enter hourly rate: "))
hours_worked = int(input("Enter hours worked: "))
tax_bracket = float(input("Enter tax bracket (e.g., 0.2): "))
gross_pay = hourly_rate * hours_worked
net_pay = gross_pay - (gross_pay * tax_bracket)
divider = "=" * 40
print(divider)
print(f"{'OFFICIAL PAYSLIP':^40}")
print(divider)
print(f"{'Hourly Rate:':<25}${hourly_rate:>14.3f}")
print(f"{'Hours Worked:':<25}${hours_worked:>14.3f}")
print(f"{'Tax Bracket:':<25}${tax_bracket:>14.3f}")
print(f"{'Gross Pay:':<25}${gross_pay:>14.3f}")
print(f"{'Net Pay:':<25}${net_pay:>14.3f}")

print(divider)
```
