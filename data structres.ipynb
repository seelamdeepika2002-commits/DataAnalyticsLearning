```python
#10th question
#Q10. Sets — Deduplicating and Comparing Survey Data
#
# day1_respondents = ["A101", "A102", "A103", "A104", "A101"]
# day2_respondents = ["A102", "A103", "A105", "A106"]
#
# Write code
# ------------------------------------------------------------
#  (converting to sets first) to print:
# - Unique respondents on Day 1
# - Respondents who answered on BOTH days
# - Respondents who answered ONLY on Day 1
# - All unique respondents across both days combined
#
# Expected output:
# {'A101', 'A102', 'A103', 'A104'}
# {'A102', 'A103'}
# {'A101', 'A104'}
# {'A101', 'A102', 'A103', 'A104', 'A105', 'A106'}
day1_respondents = ["A101", "A102", "A103", "A104", "A101"]
day2_respondents = ["A102", "A103", "A105", "A106"]

# Convert lists to sets
day1 = set(day1_respondents)
day2 = set(day2_respondents)

# Unique respondents on Day 1
print(day1)

# Respondents who answered on BOTH days
print(day1 & day2)

# Respondents who answered ONLY on Day 1
print(day1 - day2)

# All unique respondents across both days
print(day1 | day2)
```

```python
#11th question
#Q11. Sets — Building and Validating a Tag List
#
# required_tags = {"electronics", "sale", "featured"}
# product_tags = ["sale", "electronics", "sale", "new", "featured", "new"]
#
# Write code to:
# - Convert product_tags into a set of unique tags
# - Check if all required_tags are present in the product's tag set
# - Print which required tags (if any) are missing
#
# Expected output:
# {'sale', 'electronics', 'new', 'featured'}
# True
# set()
# -------------------
required_tags = {"electronics", "sale", "featured"}
product_tags = ["sale", "electronics", "sale", "new", "featured", "new"]

# Convert product_tags into a set
product_tag_set = set(product_tags)

# Check if all required tags are present
all_required = required_tags.issubset(product_tag_set)

# Find missing required tags
missing_tags = required_tags - product_tag_set

print(product_tag_set)
print(all_required)
print(missing_tags)
```
